# **Data Cleaning**

```python
#creating a copy for cleaning the dataset
sales_clean = df.copy()
duplicate_before= sales_clean.duplicated().sum()
missing_before = sales_clean.isnull().sum()
rows_before = sales_clean.shape[0]
print("Duplicate row before cleaning is =",duplicate_before)
print("Missing values before cleaning:")
print(missing_before[missing_before > 0])
print("Rows before cleaning:",rows_before)
print()

#Outlier Concept
print("="*50)
print("Detecting the Outliers")
print("="*50)
numeric_cols = df.select_dtypes(include="number").drop(columns=["Sales","Quantity","Discount"]).columns
#formula IQR
#Outlier Concept
for column in numeric_cols:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    print("-" * 40)
    print("Column=",column)
    print("Lower Bound=", lower_bound)
    print("Upper Bound=", upper_bound)
    print("Number of Outliers=", outliers.shape[0])
    print("Min value=",df[column].min())
    print("Max value=",df[column].max())
    print("-" * 40)

# dealing with quantity sales and discount because can't be in the negative values
cols=["Quantity","Sales","Discount"]
for column in cols:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    lower_bound=max(0,lower_bound)
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    print("-" * 40)
    print("Column=",column)
    print("Lower Bound=", lower_bound)
    print("Upper Bound=", upper_bound)
    print("Number of Outliers=", outliers.shape[0])
    print("Min value=",df[column].min())
    print("Max value=",df[column].max())
    print("-" * 40)

```

```python
#handling of the missing value

#Filling missing value of coustmer name with"unknown"
sales_clean["Customer Name"] = sales_clean["Customer Name"].fillna("Unkown")

# removing the missing rows of postal code
sales_clean =sales_clean.dropna(subset=["Postal Code"])

#fill the values of sales and profit
for col in ["Sales", "Profit"]:
    sales_clean[col] = sales_clean[col].fillna(sales_clean[col].median())

# Verify Missing values
print("Missing value after treatement=")
print(sales_clean.isnull().sum())
```

```python
# Removing the duplicated rows
sales_clean = sales_clean.drop_duplicates()
rows_after_duplicates = sales_clean.shape[0]

print("Duplicate rows found before cleaning:", duplicate_before)
print("Rows remaining after duplicate removal:", rows_after_duplicates)
print("Duplicate rows after cleaning:", sales_clean.duplicated().sum())
print("Duplicate Rows has been removed!")

#Check the data type
print("Data Type before Conversion")
print("="*40)
print(sales_clean[["Order Date","Ship Date"]].dtypes)

# Order Date and Ship Date after Conversion into date time format
sales_clean["Order Date"]= pd.to_datetime(sales_clean["Order Date"] , format ="mixed")
sales_clean["Ship Date"] = pd.to_datetime(sales_clean ["Ship Date"], format="mixed")
print()
print("Data Type after Conversion")
print("="*40)
print(sales_clean[["Order Date","Ship Date"]].dtypes)
print("Order Date and Ship Date has been succesfully converted into date time format")
print()

# Removing Extra Spaces
# Data Cleaning removing leading and trailing spaces and using standard captilization
text_cols = sales_clean.select_dtypes( include= ["object","string"]).columns
for col in text_cols:
  sales_clean[col] = sales_clean[col].str.strip().str.title()
print("Text Cleaning have been completed")
print(list(text_cols))
print()

# Verify the cleaned dataset
print("Cleaned Dataset Information:")
sales_clean.info()
print()

print("Missing values after cleaning:")
print(sales_clean.isnull().sum())
print()

print("Duplicate rows after cleaning:", sales_clean.duplicated().sum())
```

```python
# Data Cleaning Report

rows_after_cleaning = sales_clean.shape[0]

print("DATA CLEANING REPORT")
print("=" * 50)

print("1. Missing Value Treatment")
print("Missing columns before cleaning:")
print(missing_before[missing_before > 0])
print("Missing values were handled using fillna() and dropna().")
print()

print("2. Duplicate Removal Summary")
print("Duplicate rows found:", duplicate_before)
print("Rows before cleaning:", rows_before)
print("Rows after cleaning:", rows_after_cleaning)
print()

print("3. Data Type Corrections")
print("Order Date and Ship Date were converted to datetime.")
print("Postal Code was converted to integer after removing missing values.")
print()

print("4. Text Cleaning Summary")
print("Extra spaces were removed from text columns using str.strip().")
print()

print("Final Conclusion")
print("="*50)
print("1. Missing values have been handled.")
print("2. Duplicate rows have been removed.")
print("3. Date columns are now in datetime format.")
print("4. Text columns are cleaned.")
print("5. The dataset is ready for feature engineering and analysis.")
print("6. The data loss cleaning.",(len(df)-len(sales_clean))/len(df)*100)

```

# **Feature Engineering**

```python
# Feature Engineering
# ==========================

# Ensure 'Order Date' and 'Ship Date' are datetime objects
sales_clean["Order Date"] = pd.to_datetime(sales_clean["Order Date"], format="mixed")
sales_clean["Ship Date"] = pd.to_datetime(sales_clean["Ship Date"], format="mixed")

# 1. Create Order Year
sales_clean["Order Year"] = sales_clean["Order Date"].dt.year


# 2. Create Order Month
sales_clean["Order Month"] = sales_clean["Order Date"].dt.month_name()
sales_clean["Order Month Number"] = sales_clean["Order Date"].dt.month

# 3. Create Delivery Days
sales_clean["Delivery Days"] = (
    sales_clean["Ship Date"] - sales_clean["Order Date"]
).dt.days

# 4. Create Profit Margin
sales_clean["Profit Margin"] = (
    sales_clean["Profit"] / sales_clean["Sales"]
) * 100

# 5. Create Profit Status
sales_clean["Profit Status"] = np.where(sales_clean["Profit"] > 0,"Profit","Loss")
#print 5 rows
sales_clean.head()

```

```python
# QUESTIONS & ANSWERS
# ==========================

most_orders_year = sales_clean["Order Year"].value_counts().idxmax()

most_orders_month = sales_clean["Order Month"].value_counts().idxmax()

avg_delivery_time = sales_clean["Delivery Days"].mean()

avg_profit_margin = sales_clean["Profit Margin"].mean()

loss_making_orders = sales_clean[
    sales_clean["Profit Status"] == "Loss"
]

print("1. Year with most orders:", most_orders_year)
print("2. Month with most orders:", most_orders_month)
print("3. Average delivery time:", round(avg_delivery_time, 2), "days")
print("4. Number of loss-making orders:", loss_making_orders.shape[0])
print("5. Average profit margin:", round(avg_profit_margin, 2), "%")

print("\nLoss-making Orders Sample:")
print(
    loss_making_orders[
        ["Order ID",
         "Category",
         "Sub-Category",
         "Sales",
         "Profit",
         "Profit Status"]
    ].head()
)
```
